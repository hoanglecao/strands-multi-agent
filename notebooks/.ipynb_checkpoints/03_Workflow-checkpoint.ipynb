# Lab 3 — Workflow with Parallel Agents

A developer-defined task DAG: collect → parallel metrics/logs → report. This demonstrates the Workflow pattern via Python orchestration, **not** a native Strands Workflow orchestrator API.

**Run cells from top to bottom.** Execute `00_Setup.ipynb` first. Each notebook is intended to be opened from the `notebooks/` folder.

```python
import os, sys, json, time
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor
from datetime import datetime, timezone
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
os.chdir(ROOT)
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from strands.session import SnapshotSessionManager
from strands.storage import LocalFileStorage
from common import INCIDENT, STATE, build_agent, get_incident_metrics, get_incident_logs
```

## 1. Define workflow Agents and session

The report-writer conversation persists between invocations. The **overall Python workflow state** is checkpointed separately.

```python
metrics_agent = build_agent(
    name="metrics_analyst", description="Reads synthetic incident metrics.",
    system_prompt="Call get_incident_metrics for INC-1042. State observed metrics and uncertainty.",
    tools=[get_incident_metrics],
)
logs_agent = build_agent(
    name="logs_analyst", description="Reads synthetic incident logs.",
    system_prompt="Call get_incident_logs for INC-1042. State log evidence and uncertainties.",
    tools=[get_incident_logs],
)
report_session = SnapshotSessionManager(
    session_id="nb-report-inc1042", storage=LocalFileStorage(str(STATE / "sessions")),
)
report_agent = build_agent(
    name="report_writer", description="Combines analyses into a safety-focused incident report.",
    system_prompt="Write Evidence, Hypothesis, Recommended Human Checks, and Uncertainty. Use only supplied fresh evidence.",
    use_memory=True, session_manager=report_session,
)
print("Workflow Agents ready")
```

## 2. Stage 1 — Collect metadata (deterministic code)

```python
metadata = json.dumps({"incident_id": INCIDENT["incident_id"], "service": INCIDENT["service"]})
print(metadata)
```

## 3. Stage 2 — Parallel independent analysis

Use different Agent instances in each thread. Do not invoke the same mutable Agent instance concurrently.

```python
started = time.monotonic()
with ThreadPoolExecutor(max_workers=2) as pool:
    fm = pool.submit(metrics_agent, "Investigate INC-1042 metrics using your read-only tool.")
    fl = pool.submit(logs_agent, "Investigate INC-1042 logs using your read-only tool.")
    metrics_result = str(fm.result())
    logs_result = str(fl.result())
print("Parallel stage seconds:", round(time.monotonic()-started, 2))
print("Metrics excerpt:", metrics_result[:600])
print("Logs excerpt:", logs_result[:600])
```

## 4. Stage 3 — Join results into the report-writing Agent

```python
report_prompt = "Create a report using only fresh synthetic evidence: " + str({
    "metadata": metadata,
    "metrics": metrics_result,
    "logs": logs_result,
})
report = report_agent(report_prompt)
print(report)
```

## 5. Write a workflow-level checkpoint

```python
checkpoint = {
    "timestamp": datetime.now(timezone.utc).isoformat(), "incident": "INC-1042",
    "stages": ["collect", "metrics_and_logs_parallel", "report"],
    "report_excerpt": str(report)[:700],
}
checkpoint_path = STATE / "notebook_workflow_run.json"
checkpoint_path.write_text(json.dumps(checkpoint, indent=2), encoding="utf-8")
print("Checkpoint:", checkpoint_path)
print("Saved:", checkpoint_path.exists())
```

**Checkpoint:** Both parallel analyses finish before the report step. The report-writer session and workflow-level checkpoint are different persistence mechanisms.
