# Lab 1 — Long-Term Memory

Write a non-sensitive reporting preference and retrieve it using a new Agent instance.

**Run cells from top to bottom.** Execute `00_Setup.ipynb` first. Each notebook is intended to be opened from the `notebooks/` folder.

```python
import sys, os
from pathlib import Path
ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
os.chdir(ROOT)
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from strands import Agent
from common import make_memory_manager, MEMORY_FILE
print("Memory file:", MEMORY_FILE)
```

## 1. Write memory

This code creates a Strands Agent that saves your preferred incident report format into Memory, so the preference can be retrieved and reused in future conversations.

```python
writer = Agent(
    name="preference_writer",
    system_prompt=("Store only harmless report-format preferences when requested. "
                   "Call add_memory; confirm only after the tool reports success."),
    memory_manager=make_memory_manager(allow_writes=True),
    context_manager="auto",
    callback_handler=None
)
write_result = writer("Save my preference: incident reports should include Evidence, Hypothesis, Recommended Checks, and Uncertainty. Use add_memory.")
print(write_result)
```

## 2. Simulate a fresh session

Create a **different Agent instance**. It should search the persistent memory store rather than relying on the previous Agent history.

```python
reader = Agent(
    name="preference_reader",
    system_prompt="When asked about reporting preferences, use search_memory; never invent stored values.",
    memory_manager=make_memory_manager(allow_writes=False),
    context_manager="auto",
    callback_handler=None
)
read_result = reader("What report format do I prefer? Search memory before answering.")
print(read_result)
print("Memory storage exists:", MEMORY_FILE.exists())
if MEMORY_FILE.exists(): print("Stored bytes:", MEMORY_FILE.stat().st_size)
```
